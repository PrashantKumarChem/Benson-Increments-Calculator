<ul>
<li>This notebook is designed to calculate approximate heats of formation of organic molecules based on the idea of Benson Group Increments: Cohen & Benson, <em>Chem. Rev.</em> <b>1993</b>, <em>93</em>, 2419.</li>
<li>The notebook takes values from a user's file titled "increment_table.csv" that is in the same folder as this notebook file. This csv file contains one column titled "Benson Group Increment" populated by Benson Group increment types and a second column titled "Delta_Hf kcal/mol" populated by numerical values in kcal/mol. Research is ongoing to update values and add increments to better describe a range of molecules; the user can decide which increments to use in the calculator.</li>
<li>The Benson Group Increments are rendered into compact buttons that the user can click on to select. The value associated with that button is automatically added to the total displayed below the buttons. The user needs to decide which increments and corrections are needed based on the molecule of interest to make the entire process of estimation transparent.</li> 

```python
import pandas as pd
import ipywidgets as widgets
from IPython.display import display

# Load data from the CSV file
df = pd.read_csv('increment_correction_table.csv')

# Clean column names by stripping any leading/trailing spaces
df.columns = df.columns.str.strip()

# Ensure the CSV file has columns 'Benson Group Increments', 'Delta_Hf kcal/mol', 'Corrections', and 'Values'
# Fill NaN values with a default value (e.g., an empty string or zero)
df.fillna(0, inplace=True)

# Create dictionaries for the button values
CH_notation_values = dict(zip(df['CH Benson Group Increment'], df['Delta_Hf kJ/mol']))
CHO_notation_values = dict(zip(df['CHO Benson Group Increment'], df['CHO Values']))
correction_values = dict(zip(df['Correction'], df['Correction Values']))

# Variable to store the total value
total_value_kj = 0.0
last_value_kj = 0.0  # Store the last added value for undo purposes
pressed_buttons = []  # List to store pressed buttons

# Conversion factor from kJ to kcal
kj_to_kcal = 0.239006

# Create labels to display the standard heat of formation in kJ/mol and kcal/mol
total_label_kj = widgets.Label(value=f"Standard heat of formation: {total_value_kj:.2f} kJ/mol")
total_label_kcal = widgets.Label(value=f"Standard heat of formation: {total_value_kj * kj_to_kcal:.2f} kcal/mol")
pressed_buttons_label = widgets.Label(value="Pressed buttons: None")

# Function to update the total when a button is clicked
def add_to_total(value, label):
    global total_value_kj, last_value_kj
    total_value_kj += value
    last_value_kj = value  # Store the last value added
    pressed_buttons.append(label)  # Add the button label to the list
    total_label_kj.value = f"Standard heat of formation: {total_value_kj:.2f} kJ/mol"
    total_label_kcal.value = f"Standard heat of formation: {total_value_kj * kj_to_kcal:.2f} kcal/mol"
    pressed_buttons_label.value = f"Pressed buttons: {', '.join(pressed_buttons)}"

# Function to undo the last action
def undo_last_action(button):
    global total_value_kj, last_value_kj
    if pressed_buttons:
        total_value_kj -= last_value_kj  # Subtract the last added value
        pressed_buttons.pop()  # Remove the last pressed button
        total_label_kj.value = f"Standard heat of formation: {total_value_kj:.2f} kJ/mol"
        total_label_kcal.value = f"Standard heat of formation: {total_value_kj * kj_to_kcal:.2f} kcal/mol"
        pressed_buttons_label.value = f"Pressed buttons: {', '.join(pressed_buttons) if pressed_buttons else 'None'}"
        last_value_kj = 0.0  # Reset last_value after undoing

# Create a function to generate compact buttons and bind the correct value
def create_button(label, value, color='lightblue'):
    if label == '' or value == '':
        return None  # Skip creating a button if label or value is empty
    button = widgets.Button(description=f"{label}", layout=widgets.Layout(width="125px", height="25px", padding="0px 0px 0px 0px"))
    button.style.button_color = color
    button.on_click(lambda b: add_to_total(value, label))
    return button

# Create compact buttons for each chemical notation from the CSV file
CH_notation_buttons = [create_button(CH_notation, value) for CH_notation, value in CH_notation_values.items() if CH_notation and value]
CHO_notation_buttons = [create_button(CHO_notation, value, color='lightpink') for CHO_notation, value in CHO_notation_values.items() if CHO_notation and value]
correction_buttons = [create_button(correction, value, color='lightgreen') for correction, value in correction_values.items() if correction and value]

# Create a compact grid layout for the buttons (e.g., 5 columns)
CH_notation_layout = widgets.GridBox([btn for btn in CH_notation_buttons if btn], layout=widgets.Layout(grid_template_columns="repeat(6, 150px)", grid_gap="2px"))
CHO_notation_layout = widgets.GridBox([btn for btn in CHO_notation_buttons if btn], layout=widgets.Layout(grid_template_columns="repeat(6, 150px)", grid_gap="2px"))
correction_layout = widgets.GridBox([btn for btn in correction_buttons if btn], layout=widgets.Layout(grid_template_columns="repeat(6, 150px)", grid_gap="2px"))

# Create the Undo button to allow user to remove the last increment added to the total value
undo_button = widgets.Button(description="Undo the last addition", layout=widgets.Layout(width="900px", height="25px"))
undo_button.on_click(undo_last_action)

# Display the compact grid of clickable buttons (table), the undo button, and the total labels
display(CH_notation_layout)
display(widgets.HTML("<br>"))  # Add space break
display(CHO_notation_layout)
display(widgets.HTML("<br>"))  # Add space break
display(correction_layout)
display(undo_button)
display(total_label_kj)
display(total_label_kcal)
display(pressed_buttons_label)
```
